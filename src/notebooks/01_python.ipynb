```python
# function
def grade_risk(score: float) -> str:
    """
    Bucket a credit score into a risk level.

    :param score: A credit score, between 300 and 850.
    :return: Either 'high', 'medium' or 'low'
    """
    if score < 580:
        return 'high'
    if score > 580 and score < 670:
        return 'medium'

    return 'low'


# list of credit scores
credit_scores: list[float] = [485.0, 637.0, 815.0]

# for loop
for credit_score in credit_scores:
    print(f"{credit_score} grade risk is {grade_risk(credit_score)}")

# conditional expression
threshold: float = 670.0
label: str = "approve" if credit_scores[0] >= threshold else "in review"

print(f"The score {credit_scores[0]:.0f} is {label}")
```

```python
# function practice
import math

def standardize(values: list[float], *, ddof: int = 0) -> list[float] | str:
    """
    Rescale values to zero mean and unit standard deviation. This is the z-score
    transform, a common preprocessing step in machine learning where models that
    measure distance (k-NN, SVM, k-means) or that use gradient descent converge
    badly when one feature is measured in dollars and another in years.

    :param values: The sample to rescale. Must contain at least two elements.
    :param ddof: Delta degrees of freedom. 0 for population std, 1 for sample estimate.
    :return: A new list of the same length, with mean 0 and std 1.
    """
    if len(values) < 2:
        message: str = f"Need at least 2 values to standardize, provided {len(values)}"

        return message

    mean: float = sum(values) / len(values)
    variance: float = sum((value - mean) ** 2 for value in values) / (len(values) - ddof)
    std: float = math.sqrt(variance)

    if std == 0.0:
        message: str = "Cannot standardize. All values are identical."

        return message

    return [(value - mean) / std for value in values]

sample_credit_scores: list[float] = [637.0, 815.0, 723.0, 716.0, 675.0, 488.0]
z_score: list[float] = standardize(sample_credit_scores)

print("credit scores: ", sample_credit_scores)
print("z-score: ", [round(value, 3) for value in z_score])
print(f"mean {sum(z_score) / len(z_score):.1e}, std {math.sqrt(sum(value ** 2 for value in z_score) / len(z_score)):.4f}")
```

```python
from typing import Self, Sequence
import math

# classes
class RunningStats:
    """
    Streaming mean and variance, computed in a single pass.
    """

    # class attribute shared by every instance
    algorithm: str = "welford"

    def __init__(self) -> None:
        """
        Start with an empty accumulator.
        """
        self.count: int = 0
        self._mean: float = 0.0
        self._m2: float = 0.0   # sum of squared deviations from the running mean

    def update(self, value: float) -> Self:
        """
        Incorporate one observation.

        :param value: The observation.
        :return: ``self``, so updates can be chained.
        """
        self.count += 1
        delta: float = value - self._mean
        self._mean += delta / self.count
        self._m2 += delta * (value - self._mean)

        return self

    @property
    def mean(self) -> float:
        """
        The running arithmetic mean.

        :return: The mean, otherwise NaN if not data
        """
        return self._mean if self.count else math.nan

    @property
    def variance(self) -> float:
        """
        The sample variance.

        :return: The variance, other NaN if fewer than 2 values
        """
        return self._m2 / (self.count - 1) if self.count > 1 else math.nan

    @property
    def std(self) -> float:
        """
        The sample standard deviation.

        :return: The sample standard deviation
        """
        return math.sqrt(self.variance) if self.count > 1 else math.nan

    @classmethod
    def from_values(cls, values: Sequence[float]) -> Self:
        """
        Build an accumulator already fed with ``values``.

        :param values: Observations to consume immediately.
        :return: A populated accumulator.
        """
        stats = cls()

        for value in values:
            stats.update(value)

        return stats

    @staticmethod
    def is_outlier(value: float, mean: float, std: float, *, z: float = 3.0) -> bool:
        """
        Whether a value lies more than ``z`` standard deviations from the mean.

        :param value: The observation to test.
        :param mean: Reference mean.
        :param std: Reference standard deviation.
        :param z: Threshold in standard deviation.
        :return: ``True`` if the value is beyond the threshold
        """
        return std > 0 and abs(value - mean) / std > z

    # dunder methods
    def __repr__(self) -> str:
        return f"{type(self).__name__} (count={self.count}, mean={self._mean:.4f}, std={self.std:.4f})"

    def __len__(self) -> int:
        return self.count

    def __bool__(self) -> bool:
        return self.count > 0


stats: RunningStats = RunningStats.from_values([637.0, 815.0, 723.0, 716.0, 675.0, 488.0])
print(stats)
print(f"len(stats) = {len(stats)}, bool(stats) = {bool(stats)}")
print(f"mean = {stats.mean:.2f}, std = {stats.std:.2f}")
print(f"is 1200 an outlier? {RunningStats.is_outlier(1200.0, stats.mean, stats.std)}")
print(f"chaining: {stats.update(700.0).update(710.0)}")
```
